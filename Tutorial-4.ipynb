# Array Operators

```python
import numpy as np
```

```python
# Assign the array

a = np.arange(12).reshape(3,4)
b = np.arange(1,13).reshape(4,3)

print(b)
```

```python
# Scalar Operation

# Arithematic operator - Addition

plus = 2 + a
print("Adition\n",plus) 
```

```python
# Arithematic - Exponential

expo = 2 ** b
print("Exponential\n",expo)
```

```python
# Relational 

# greater

greater = b > 5
print(greater)
```

```python
# Equal to

equal = (a == 10)
print(equal)
```

```python
# Vector operations

# Arithmetic - Addition

add = a + (b.reshape(3,4))

print(add)
```

```python
# arithmetic - subtration

sub = a.reshape(4,3) - b

sub
```
