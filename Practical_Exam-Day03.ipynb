# What is the difference between `np.sum()` and `np.cumsum()`?

` np.sum()`:

- Computes the total sum of elements

- Returns one value per axis (or a single scalar)

`np.cumsum()`:

- Computes the running (cumulative) sum

- Each element includes the sum of all previous elements

- Returns an array of the same shape

```python
arr = np.array([1,2,3,4,5,6,7])
arr
```

```python
# np.sum()
arr.sum()
```

```python
# np.cumsum() -> Cumlative sum --> adding previous number with present (e.g. fibonacci)

arr.cumsum()
```

# Normalize a NumPy array using Min-Max normalization.

```python
arr = np.array([10,20,30,40,50])
arr
```

### norm = X - X(min) / X(max) - X(min) --> Formula

```python
val_min = arr.min()
val_max = arr.max()

norm = (arr-val_min) / (val_max-val_min)
norm
```

# Perform Z-score normalization on a dataset.

```python
data = np.array([10, 20, 30, 40, 50])


mean = np.mean(data)
sd = np.std(data)

z = data - mean / sd
print("Z-score normalization:",z)
```
