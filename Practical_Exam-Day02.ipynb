# Explain boolean indexing and its importance in data analysis.

```python
c = np.random.randint(1,100,25).reshape(5,5)
c
```

#### Boolean indexing is a technique used to filter, select, or modify data based on logical conditions (True / False). It is widely used in data analysis libraries such as Pandas and NumPy to work with subsets of data efficiently.

```python
[c>50]
```

# Select all values greater than the mean of an array.

```python
m = np.mean(c)
print("mean:",m)
print("Values that are greater than mean:\n",c[c > m])
```

# What is fancy indexing and how is it different from slicing?


- Fancy indexing in NumPy is a way to access array elements using an array or list of indices (or a boolean array), allowing selection of non-contiguous, arbitrary elements.

- Slicing, on the other hand, selects continuous ranges of elements using start:stop:step.

```python
arr = np.array([10, 20, 30, 40, 50])

print("Slicing:",arr[1:4])        # Slicing → [20 30 40]
print("Fancy Indexing:",arr[[0, 2, 4]])  # Fancy indexing → [10 30 50]
```

# Demonstrate the difference between a view and a copy in NumPy.

- A view does not create new data. It only provides a different way to look at the same underlying array.

```python
arr = np.array([10, 20, 30, 40, 50])

view_arr = arr[1:4]   # slicing → view
view_arr[0] = 999

print(arr)
print(view_arr)
```

# How do you apply multiple conditions when filtering a NumPy array?

```python
arr = np.random.randint(-10,10,24).reshape(4,6)
arr
```

```python
result = arr[(arr>0) & (arr%2==0)]
result
```
