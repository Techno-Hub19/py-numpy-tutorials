# Why is NumPy significantly faster than Python lists for numerical computation?

- It has contigious memory storage - better CPU cache utilization
- Broadcasting is supported, which avoids creating unnecessary temporary array
- Uses homogeneous data types, with fixed sized 
- Supports vectorization, processing entire arrays at once
- Enables broadcasting without copying data
- Allows us the perform high level mathematics operations in ease

# Explain the difference between `np.array()` and `np.asarray()`.
- `np.array()`: Always creates the new numpy array and (copy by default)
- Even if the input value is numpy array  
- It avoids modifying the original data 
- `np.asarray()`: Does not copy data if the input value is numpy array
- Creates view/reference when ever possible
- Faster and more-memory efficient

# How does NumPy store data in memory, and why does this matter for performance?
- NumPy arrays store elements in "contigeous" block of memory
- All elements are homogeneous(same) data type
- Better CPU cache utilization, where cache is store in lines ~ 64 bytes
- Supports views instead of copies - Multiple arrays can point to the same data.
- Uses row-major order by default - row operations are very fast
- Enables vectorized operations - operations are performed on complete array at once

# What is the purpose of NumPy data types (`dtype`)?
- dtype is nothing but the data type, it shows the data type of array
- Using the dtype, we can change the memory size of data type as well 
- How much memory each element uses

# Convert a float array into an integer array and analyze the impact on memory

```python
a = np.arange(12,dtype=float).reshape(3,4)
a.itemsize
```

```python
b = a.astype(np.dtype('int16'))
b.itemsize
```

# When should you explicitly specify the `dtype` while creating a NumPy array?
- To control memory usage, Default integers/floats may use more memory than needed.
- To avoid unintended type inference, NumPy may infer an unexpected dtype.
- To improve performance, Smaller, fixed-size types fit better in CPU cache.
- When working with large datasets, Even small dtype changes significantly reduce memory.
