# Array Attribute

```python
import numpy as np
```

```python
a= np.array([12,3,4,5,6])
print(a)
print(type(a))
```

```python
a = np.arange(10)
b = np.arange(12,dtype=float).reshape(4,3)
c = np.arange(8).reshape(2,2,2)
```

```python
a #1D array
```

```python
b # 2D / matrix array
```

```python
c # tensor
```

```python
# ndim - number of dimensions

print(f'Number of dimension: {a.ndim}')
```

```python
# shape

print("Array",c)
print("Shape of array",c.shape)
```

```python
# size 

print(b.size)
print(c.size)
```

```python
# itemsize - Occupied size in memory

print(b.itemsize)
```

```python
# dtype - type of data

print(a.dtype)
print(b.dtype)
```
