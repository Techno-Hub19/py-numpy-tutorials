### Broadcasting

#### - Broadcasting describes that how numpy treats arrays with different shapes and arithematic operations
#### - Here, smaller array is 'broadcasted' with larger array for compatiable shape
 

```python
import numpy as np
```

```python
a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)

print(a+b)      # Both have equal shapes of array for arithematic operationsb
```

```python
# Diff shape

a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)

print(a+b)
```

### Broadcasting Rules:

#### 1. Ensure that two arrays have same number of dimensions.
##### - If the numbers of dimensions of the two arrays are different, add new dimensions with size 1 to the head of the array with the smaller dimensions.
#### 2. Make each dimensions of the two arrays the same size.
##### - If the sizes of each dimensions of the arrays do not match, dimensions with size 1 are stretched to the size of the other array.
##### - If there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted and an error is raised.

```python
# Examples

a = np.arange(12).reshape(4,3)
b = np.arange(3)

print(a)
print(b)

print(a+b) #broadcasting is done
```

```python
a = np.arange(12).reshape(3,4)
b = np.arange(3)

print(a+b) # broadcasting will not happen
```

```python
a = np.arange(3).reshape(1,3)
b = np.arange(3).reshape(3,1)

print(a+b)
```

```python
a = np.arange(3).reshape(1,3)
b = np.arange(4).reshape(4,1)

print(a+b)
```

```python
a = np.array([1])
# shape (1,1)
b = np.arange(4).reshape(2,2)

print(a+b)
```

```python
a = np.arange(12).reshape(3,4)
b = np.arange(12).reshape(4,3)


print(a+b)  # Error will occur bcoz 1 is not there, we cannot have broadcasting
```

```python
a = np.arange(16).reshape(4,4)
b = np.arange(4).reshape(2,2)

print(a+b) # shape is different 
```
