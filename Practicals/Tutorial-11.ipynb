# Numpy array v/s Python List

```python
# speed 
# list

a = [i for i in range(100000000)]
b = [i for i in range(100000000,200000000)]

import time 

start = time.time()
for i in range(len(a)):
    c = (a[i]+b[i])
print(time.time()-start)
```

```python
# numpy

a = np.arange(100000000)
b = np.arange(100000000,200000000)

start = time.time()

c = a + b

print(time.time()-start)
```

```python
# memory 
# list

lst = [i for i in range(10000000)]

import sys

sys.getsizeof(lst)
```

```python
# array

a = np.arange(10000000,dtype=np.int8)

sys.getsizeof(a)
```
