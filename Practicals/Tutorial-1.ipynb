## What is Numpy ?
##### - Numpy is scientific computing library of python.
##### - Numpy is core library for Pandas, SciPy, TensorFlow and etc.
##### - Numpy arrays are convinent for maths calculations.
##### - It consist of homogenous(same) data types  

------------------------------------------------------------------------------------------------------------------
## Difference between Sequence & Array
##### - Array consumes less memory space & store large amount of data..
##### - Array can perform mathematical calculations faster as compared to sequence.
##### - Numpy arrays are fixed in size, list are dynamic in size.

```python
# import library
import numpy as np
```

```python
# np.array

a = np.array([1,2,3,4])     # 1D array
print(type(a))
a
```

```python
# 2D & 3D

a = np.array([[1,2,3,4],[5,6,7,8]])   # [[]] - defines the 2d array
print("2D array:\n",a)


b = np.array([[[1,2,],[3,4]],[[5,6],[7,8]]])     # [[[]]] - defines the 3d array 
print("3D array:\n",b)
```

```python
#dtype

a = np.array([1,2,3,4,5,6,7])
a.dtype     # data type of array
```

```python
a = np.array([1,2,3,4],dtype=complex) # complex
b = np.array([1,2,3,4],dtype=float) # float
c = np.array([1,2,3,4],dtype=bool) # bool

print("complex:",a)
print("float:",b)
print("bool:",c)
```

```python
# arange - It gives random number from given range

a = np.arange(1,5)
print(a)
b = np.arange(1,5,2)
print(b)
```

```python
# With reshape - Product should be equal to number of items in array 

np.arange(1,11).reshape(5,2)
```

```python
# np.ones and np.zeros - Used in nueral networks

np.ones((3,3))
```

```python
np.zeros((3,3))
```

```python
# np.random

np.random.random((3,3))
```

```python
# np.linspace -linearly space/ linear space

a = np.linspace(1,10,15)  # 3rd index shows how many number we want
print(a,'\n')

b = np.linspace(1,10,10,dtype=int)
print(b)
```

```python
# np.identity - used for identity matrix

np.identity(3)  # diagonal items are 1
```
