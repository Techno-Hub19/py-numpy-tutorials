# Array Functions

```python
import numpy as np
```

```python
a = np.random.random((3,3))
a = np.round(a*100)
a
```

```python
# max

np.max(a)
```

```python
# min

np.min(a)
```

```python
# sum

np.sum(a)
```

```python
# prod

np.prod(a)
```

```python
# col - 0, row - 1

print('row-wise:',np.max(a,axis=1))   
print('col-wise:',np.min(a,axis=0))
```

```python
# mean

np.mean(a)
```

```python
# median

np.median(a)
```

```python
# std

np.std(a)
```

```python
# var

np.var(a)
```

```python
# trigonomteric function - not used in data science

np.sin(a)
```

```python
# dot product

a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12).reshape(4,3)

np.dot(a1,a2)
```

```python
# log 

c = np.log(a)

c
```

```python
# exponent

np.exp(a)
```

```python
# round

np.round(c)
```

```python
# floor

np.floor(c)
```

```python
# ceil

np.ceil(c)
```
