# Indexing

```python
import numpy as np
```

```python
a = np.arange(13)
a
```

### Note: [row,col] for indexing and slicing

```python
# indexing

a[-1]
```

```python
b = np.arange(12).reshape(3,4)
b
```

```python
# 2D array indexing - [row,index]

b[0,1]
```

```python
b[2,3]
```

```python
# 3D - Tensor - [arr,row,index]

c = np.arange(8).reshape(2,2,2)
c
```

```python
c[0,1,1]
```

```python
# get the 6 from the tensor

c[1,1,0]
```

# Slicing

```python
a
```

```python
# slicing

a[:]
```

```python
a[2:5]
```

```python
a[0:13:2]
```

```python
# 2D- slicing
b
```

```python
b[:,0]      #[row,col]
```

```python
b[:2,1:]
```

```python
b[1,::3]
```

```python
b[1,1:3]
```

```python
b[::2,1::2]
```

```python
b[1:,1:3]
```

```python
b[::2,::3]
```

```python
# 3D - slicing

d = np.arange(27).reshape(3,3,3)
d
```

```python
d[1]
```

```python
d[::2,0,::2]
```

```python
d[2,1:,1:]
```

```python
d[1,:,1]
```

```python
d[0,1]
```

```python
d[::2]
```
