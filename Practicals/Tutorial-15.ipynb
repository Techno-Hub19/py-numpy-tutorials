### np.sort

```python
import numpy as np
```

```python
a = np.random.randint(1,50,25)
a
```

### np.append
#### - appends the value at the end of the array

```python
np.append(a,100)
```

```python
np.append(b,np.ones((b.shape[0],1)),axis=1)

# b.shape is --> row, 1 --> add one
```

```python
b.shape
```

```python
np.append(b,np.zeros((1,b.shape[1])),axis=0)

# b.shape is --> col, 0 --> add one
# shape[2] --> depth (3D arrays)
```

### Set Functions

```python
a = np.array([1,2,3,4,5])
b = np.array([6,2,3,9,10])
```

```python
# union1d

np.union1d(a,b)     # all elements without duplicates
```

```python
# intersect1d

np.intersect1d(a,b)     # common elements from both intersects
```

```python
# setdiff1d

np.setdiff1d(a,b)
```

```python
np.setdiff1d(b,a)
```

```python
# np.setxor1d

np.setxor1d(a,b)    # removes common element
```
