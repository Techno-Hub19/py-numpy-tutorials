# Advance Indexing

```python
import numpy as np
```

```python
# Normal indexing and slicing

a = np.arange(24).reshape(6,4)

a
```

```python
# Indexing - [row,col]
a[1,2]
```

```python
# Slicing [row:row,col:col]
a[1:,2:]
```

```python
# Fancy Indexing
a
```

```python
# fancyy indexing for row
a[[0,2,3,5]]
```

```python
# fancy indexing for column

a[:,[0,1,3]]
```

```python
# Boolean Indexing

b = np.random.randint(1,100,24,dtype=np.int16).reshape(6,4)
b
```

```python
# Find all numbers greater than 50.

a[a>50]
```

```python
# Find out all even numbers

a[a%2==0]
```

```python
# Find all numbers greater than 50 and even numbers

a[(a>50)&(a%2==0)]
```

```python
# Find all number that are divisible by 5

a[(a%2 == 0)]
```

```python
# Find all numbers not divisible by 7

a[~(a%7 == 0)]
```
