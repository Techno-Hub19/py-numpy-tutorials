# Iterating

```python
import numpy as np
```

```python
a = np.arange(12)
a
```

```python
for a in a:
    print(a)
```

```python
b = np.arange(12).reshape(3,4)
b
```

```python
for b in b:
    print(b)
```

```python
c = np.arange(27).reshape(3,3,3)
c
```

```python
for i in c:
    print(i)
```

```python
for c in np.nditer(c):
    print(c)
```
