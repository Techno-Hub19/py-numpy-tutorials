## PROJECT:  Sales Data Analytics using python (Numpy)

This project is based on sales data of random store were we're trying to analyze the data and get some helpful insights from it. Which will help us to get the idea about the actual sales and performance. We will be finding it by questions and resolving with numpy that will reduce our calculation time and will give accurate and expected output from it.

```python
# import numpy lib 
import numpy as np
```

```python
sales_data = np.array([
    [31218, 45539, 90900, 45726],   # Jan
    [53421, 36402, 60840, 69007],   # Feb
    [75231, 64721, 71131, 22826],   # Mar
    [20020, 15226, 64582, 55445],   # Apr
    [20162, 70325, 85903, 24611],   # May
    ])
sales_data
```

```python
# Total sum of products

total_price = np.sum(sales_data)
print("Total sum of sales",total_price)
```

```python
# Month wise total sales

total_sales_monthly = np.sum(sales_data,axis=1)
print("Monthly total sales",total_sales_monthly)
```

```python
# Average sales of Products 

avg_sales = np.average(sales_data)
print("Avg price of sales data",avg_sales)
```

```python
# Month wise highest sales 

month_wise_highest_sales = np.max(sales_data,axis=1)
print("Month wise highest sales from JAN - MAY:",month_wise_highest_sales)
```

```python
# Month wise lowest sales
month_wise_lowest_sales = np.min(sales_data,axis=1)
print('Month wise lowest sales from JAN - MAY',month_wise_lowest_sales)
```

```python
# Month-to-Month Change (Difference)

growth = np.diff(sales_data)
growth
```

### Conclusion: 
This NumPy project converts raw sales numbers into business insights using statistical aggregation,  and time-series growth tracking.
